```python
import matplotlib.pyplot as plt
import numpy
import seaborn as sns
import pandas as pd
```

```python
data1 = pd.read_csv('data1.csv', encoding = 'euc-kr')
code = pd.read_excel('code.xlsx')
```

```python
data1
```

```python
dict = code[['읍면동', 'full_name']].set_index('읍면동').T.to_dict('index')
dict
```

```python
dongName = dict['full_name']
dongName
```

```python
# head = next(data1)
data1['출발지 코드(start_place_cd)'].map(dongName)
```

```python
data1['출발지 코드(start_place_cd)'] = data1['출발지 코드(start_place_cd)'].map(dongName)
```

```python
data1.groupby(['출발일시(start_dt)','연령대(agegrd_nm)'])['인구수(popl_cnt)'].sum().reset_index()
```

```python
pivot = data1.pivot_table(
    values='인구수(popl_cnt)',
    index='출발일시(start_dt)',
    columns='연령대(agegrd_nm)',
    aggfunc='sum',
    fill_value=0
)

print(pivot)
```
