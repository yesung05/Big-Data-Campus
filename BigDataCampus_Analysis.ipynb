```python
import matplotlib.pyplot as plt

# 업로드된 파일 경로
bycl_raw = "/content/datas/bycl/TL_BCYCL_INFR_ST_2019.shp"
train_raw = "/content/datas/train/TB_O_SB_STATN_2019.shp"
fclty_raw = "/content/datas/fclty/TL_TBVIATR_FCLTY_INFO_2019.shp"
map_raw = "/content/datas/Map/EMD.shp"
# SHP 파일 읽기
bycl = gpd.read_file(bycl_raw)
train = gpd.read_file(train_raw)
fclty = gpd.read_file(fclty_raw)
map = gpd.read_file(map_raw)

tm1_crs = 'PROJCS["PCS_ITRF2000_TM_1",GEOGCS["ITRF2000",DATUM["International_Terrestrial_Reference_Frame_2000",SPHEROID["GRS 1980",6378137,298.257222101,AUTHORITY["EPSG","7019"]],AUTHORITY["EPSG","6656"]],PRIMEM["Greenwich",0],UNIT["Degree",0.0174532925199433]],PROJECTION["Transverse_Mercator"],PARAMETER["latitude_of_origin",38],PARAMETER["central_meridian",127],PARAMETER["scale_factor",1],PARAMETER["false_easting",200000],PARAMETER["false_northing",600000],UNIT["metre",1,AUTHORITY["EPSG","9001"]],AXIS["Easting",EAST],AXIS["Northing",NORTH]]'

map = map.to_crs(tm1_crs)

# 시각화
fig, ax = plt.subplots(figsize=(15, 15))
map.plot(ax=ax, edgecolor='black',linewidth=0.1, color='white')
fclty.plot(ax=ax, color="orange", alpha = 0.7, markersize=7)
bycl.plot(ax=ax, color="green", markersize=1, alpha = 0.7)
train.plot(ax=ax, color="blue", alpha = 0.7, markersize=100)
ax.set_title("자전거 편의시설/지하철/주요시설 위치 시각화", fontsize=14)
plt.tight_layout()
ax.set_xlim(180000, 220000)
ax.set_ylim(535000, 567000)
# 시각화 출력 및 데이터 요약 반환
plt.show()
# gdf_info
```

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point

print(gdf.crs)
print(bycl.crs)
orig_crs = 'PROJCS["PCS_ITRF2000_TM",GEOGCS["ITRF2000",DATUM["International_Terrestrial_Reference_Frame_2000",SPHEROID["GRS 1980",6378137,298.257222101,AUTHORITY["EPSG","7019"]],AUTHORITY["EPSG","6656"]],PRIMEM["Greenwich",0],UNIT["Degree",0.0174532925199433]],PROJECTION["Transverse_Mercator"],PARAMETER["latitude_of_origin",38],PARAMETER["central_meridian",127.5],PARAMETER["scale_factor",0.9996],PARAMETER["false_easting",1000000],PARAMETER["false_northing",2000000],UNIT["metre",1,AUTHORITY["EPSG","9001"]],AXIS["Easting",EAST],AXIS["Northing",NORTH]]'
chg_crs = 'PROJCS["PCS_ITRF2000_TM_1",GEOGCS["ITRF2000",DATUM["International_Terrestrial_Reference_Frame_2000",SPHEROID["GRS 1980",6378137,298.257222101,AUTHORITY["EPSG","7019"]],AUTHORITY["EPSG","6656"]],PRIMEM["Greenwich",0],UNIT["Degree",0.0174532925199433]],PROJECTION["Transverse_Mercator"],PARAMETER["latitude_of_origin",38],PARAMETER["central_meridian",127],PARAMETER["scale_factor",1],PARAMETER["false_easting",200000],PARAMETER["false_northing",600000],UNIT["metre",1,AUTHORITY["EPSG","9001"]],AXIS["Easting",EAST],AXIS["Northing",NORTH]]'

marketXY = gpd.GeoDataFrame(
    marketXY,
    geometry=[Point(xy) for xy in zip(marketXY.X, marketXY.Y)],
    crs=orig_crs
)
marketXY = gdf.to_crs(chg_crs)
# print(marketXY.crs)
marketXY['X'] = marketXY.geometry.x
marketXY['Y'] = marketXY.geometry.y

# marketXY = marketXY[['시장명', 'X', 'Y']]

fig, ax = plt.subplots(figsize=(15, 15))
map.plot(ax=ax, edgecolor='black',linewidth=0.1, color='white')
fclty.plot(ax=ax, color="orange", alpha = 0.7, markersize=7, label = "주요시설")
bycl.plot(ax=ax, color="green", markersize=1, alpha = 0.7, label = "자전거 편의시설")
train.plot(ax=ax, color="blue", alpha = 0.7, markersize=100, label = "전철 역")
marketXY.plot(ax=ax, color="red", alpha = 0.7, markersize=100, label = "전통시장")
ax.set_title("자전거 편의시설/지하철/주요시설 위치 시각화", fontsize=14)
plt.tight_layout()
ax.set_xlim(180000, 220000)
ax.set_ylim(535000, 567000)

# marketXY.plot()
plt.legend()
plt.show()
```
