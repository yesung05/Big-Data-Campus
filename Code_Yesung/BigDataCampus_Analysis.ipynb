```python
from shapely.geometry import Point

NEAR_DIST = 1000
SEL_MARKET_IDX = 34


# 기준 시장 선택 (예: 첫 번째 시장)
market_point = market.geometry.iloc[SEL_MARKET_IDX]

# 1.5km(1500m) 버퍼 생성
buffer = market_point.buffer(NEAR_DIST)

# 시설별 필터링
train_near = train[train.geometry.within(buffer)]
bycl_near = bycl[bycl.geometry.within(buffer)]
fclty_near = fclty[fclty.geometry.within(buffer)]

# 결과 확인
print(marketXY['name0'].iloc[SEL_MARKET_IDX])
print("전철역:", train_near)
print("자전거 편의시설:", bycl_near)
print("주요시설:", fclty_near)
```

```python
import numpy as np

# 기준 시장 선택 (예: 430번째 시장)
market_x = market.geometry.iloc[SEL_MARKET_IDX].x
market_y = market.geometry.iloc[SEL_MARKET_IDX].y

# 전철역 거리 계산 및 필터링
trainXY['distance'] = np.sqrt((trainXY['X'] - market_x)**2 + (trainXY['Y'] - market_y)**2)
trainXY_in_buffer = trainXY[trainXY['distance'] <= NEAR_DIST]

# 자전거 편의시설 거리 계산 및 필터링
byclXY['distance'] = np.sqrt((byclXY['X'] - market_x)**2 + (byclXY['Y'] - market_y)**2)
byclXY_in_buffer = byclXY[byclXY['distance'] <= NEAR_DIST]

# 주요시설 거리 계산 및 필터링
fcltyXY['distance'] = np.sqrt((fcltyXY['X'] - market_x)**2 + (fcltyXY['Y'] - market_y)**2)
fcltyXY_in_buffer = fcltyXY[fcltyXY['distance'] <= NEAR_DIST]

# 결과 확인
print(trainXY_in_buffer)
print(byclXY_in_buffer)
print(fcltyXY_in_buffer)
```

```python
import numpy as np

def bycl_score_func(distance):
    # 100m에서 1, 1000m에서 약 0.05가 되도록 scale 조정
    scale = 120  # np.exp(-100/260) ≈ 0.68, np.exp(-1000/260) ≈ 0.02
    return np.exp(-distance / scale)

def train_score_func(distance):
    # 0~1000m: 선형 반비례, 1000m 이상은 0점
    return max(0, 1 - distance/1000)

# 자전거 편의시설 점수 합산
bycl_score = byclXY_in_buffer['distance'].apply(bycl_score_func).sum()

# 지하철역 점수 합산
train_score = trainXY_in_buffer['distance'].apply(train_score_func).sum()

# 주요시설 점수 (예시: 선형 반비례)
fclty_score = fcltyXY_in_buffer['distance'].apply(bycl_score_func).sum()

# 최종 접근성 점수
access_score = train_score + bycl_score + fclty_score
print("시장 접근성 점수:", access_score)

print(train_score)
print(bycl_score)
print(fclty_score)
```

```python
# 100점 만점에서 각 항목별 가중치 설정
W_TRAIN = 0.4   # 지하철역 40%
W_BYCL = 0.3    # 자전거 편의시설 30%
W_FCLTY = 0.3   # 주요시설 30%

# 항목별 점수 합산 (정규화 없이 가중치만 적용)
total_score = (train_score * W_TRAIN) + (bycl_score * W_BYCL) + (fclty_score * W_FCLTY)

# 100점 만점으로 환산 (필요시)
print("시장 접근성 점수(100점 만점):", total_score)
```

```python
import numpy as np
import pandas as pd

# 점수 함수 정의
def bycl_score_func(distance):
    scale = 120  # 300m 부근에서 급격히 감소
    return np.exp(-distance / scale)

def train_score_func(distance):
    return max(0, 1 - distance/1000)

def fclty_score_func(distance):
    # 주요시설도 자전거와 동일하게 적용 (원하면 변경 가능)
    return np.exp(-distance / scale)

# 가중치 설정
W_TRAIN = 0.4
W_BYCL = 0.3
W_FCLTY = 0.3

scores = []
for idx, row in market.iterrows():
    market_x = row['X']
    market_y = row['Y']

    # 각 시설 거리 계산
    trainXY['distance'] = np.sqrt((trainXY['X'] - market_x)**2 + (trainXY['Y'] - market_y)**2)
    byclXY['distance'] = np.sqrt((byclXY['X'] - market_x)**2 + (byclXY['Y'] - market_y)**2)
    fcltyXY['distance'] = np.sqrt((fcltyXY['X'] - market_x)**2 + (fcltyXY['Y'] - market_y)**2)

    # 버퍼 내 시설만 점수 계산
    trainXY_in_buffer = trainXY[trainXY['distance'] <= 1000]
    byclXY_in_buffer = byclXY[byclXY['distance'] <= 1000]
    fcltyXY_in_buffer = fcltyXY[fcltyXY['distance'] <= 1000]

    train_score = trainXY_in_buffer['distance'].apply(train_score_func).sum()
    bycl_score = byclXY_in_buffer['distance'].apply(bycl_score_func).sum()
    fclty_score = fcltyXY_in_buffer['distance'].apply(bycl_score_func).sum()

    total_score = (train_score * W_TRAIN) + (bycl_score * W_BYCL) + (fclty_score * W_FCLTY)

    scores.append({
        '시장명': row['시장명'],
        'X': market_x,
        'Y': market_y,
        '점수': total_score
    })

# 데이터프레임 생성
market_score_df = pd.DataFrame(scores)
print(market_score_df.head())
```

```python
fig, ax = plt.subplots(figsize=(15, 15))
map.plot(ax=ax, edgecolor='black',linewidth=0.1, color='white')
fclty.plot(ax=ax, color="orange", alpha = 0.7, markersize=5, label = "주요 집객 시설")
bycl.plot(ax=ax, color="green", markersize=1, alpha = 0.7, label = "자전거 편의시설")
train.plot(ax=ax, color="red", alpha = 0.9, markersize=20, label = "전철 역")
scatter = ax.scatter(
    market_score_df['X'],
    market_score_df['Y'],
    c=market_score_df['점수'],
    cmap='cool',
    s=70,
    alpha=0.7,
    label="전통시장/상가"
)
ax.set_title("자전거 편의시설/지하철/주요 집객 시설 및 시장 점수위치 시각화", fontsize=14)
plt.colorbar(scatter, ax=ax, label='시장 접근성 점수')
plt.tight_layout()
ax.set_xlim(180000, 220000)
ax.set_ylim(535000, 567000)
plt.legend()
plt.show()
```
